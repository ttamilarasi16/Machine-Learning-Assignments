```python
import pandas
```

```python
import pandas as pd
```

```python
dataset = pd.read_csv("insurance_pre.csv")
```

```python
dataset = pd.get_dummies(dataset, drop_first = True)
```

```python
dataset
```

```python
independent_var = dataset[["age","bmi","children","sex_male","smoker_yes"]]
```

```python
dependent_var = dataset[["charges"]]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(independent_var,dependent_var, test_size = 0.30, random_state = 0)
```

```python
from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor(criterion='poisson',max_features='sqrt',splitter='best')
model.fit(X_train,Y_train)
```

```python
import matplotlib.pyplot as plt
from sklearn import tree
tree.plot_tree(model)
plt.show
```

```python
Y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_value = r2_score(Y_test, Y_pred)
```
