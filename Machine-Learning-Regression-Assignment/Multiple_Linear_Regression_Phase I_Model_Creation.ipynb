```python
import pandas as pd
```

```python
dataset = pd.read_csv("insurance_pre.csv")
```

```python
dataset = pd.get_dummies(dataset, drop_first=True)
dataset
```

```python
independent_var = dataset[["age", "bmi", "children", "sex_male", "smoker_yes"]]
```

```python
dependent_var = dataset[["charges"]]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(independent_var, dependent_var, test_size = 0.3, random_state = 0)
```

```python
from sklearn.linear_model import LinearRegression
Model = LinearRegression()
Model.fit(X_train, Y_train)
```

```python
weight = Model.coef_
weight
```

```python
bias = Model.intercept_
bias
```

```python
from sklearn.metrics import r2_score
Y_pred = Model.predict(X_test)
r2_score_value = r2_score(Y_test, Y_pred)
r2_score_value
```

```python
import pickle
filename = "finalized_model_MLR.sav"
```

```python
pickle.dump(Model, open(filename,'wb'))
```
